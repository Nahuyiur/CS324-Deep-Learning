```python
import matplotlib.pyplot as plt
from train_mlp_numpy import train, DNN_HIDDEN_UNITS_DEFAULT, LEARNING_RATE_DEFAULT, MAX_EPOCHS_DEFAULT, EVAL_FREQ_DEFAULT

# Train with default parameters and get history
history = train(
    dnn_hidden_units=DNN_HIDDEN_UNITS_DEFAULT,
    learning_rate=LEARNING_RATE_DEFAULT,
    max_steps=MAX_EPOCHS_DEFAULT,
    eval_freq=EVAL_FREQ_DEFAULT,
    return_history=True
)

# Plot accuracy and loss curves
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(history['steps'], history['train_acc'], label='Train Accuracy', marker='o', markersize=3)
ax1.plot(history['steps'], history['test_acc'], label='Test Accuracy', marker='s', markersize=3)
ax1.set_xlabel('Training Steps')
ax1.set_ylabel('Accuracy (%)')
ax1.set_title('Training and Test Accuracy (Batch GD)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(history['steps'], history['train_loss'], label='Train Loss', marker='o', markersize=3)
ax2.plot(history['steps'], history['test_loss'], label='Test Loss', marker='s', markersize=3)
ax2.set_xlabel('Training Steps')
ax2.set_ylabel('Loss')
ax2.set_title('Training and Test Loss (Batch GD)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print experiment parameters and final results
print(f"\n{'='*60}")
print(f"Experiment Parameters (Default - Batch GD):")
print(f"  Hidden Units: {DNN_HIDDEN_UNITS_DEFAULT}")
print(f"  Learning Rate: {LEARNING_RATE_DEFAULT}")
print(f"  Max Steps: {MAX_EPOCHS_DEFAULT}")
print(f"  Eval Frequency: {EVAL_FREQ_DEFAULT}")
print(f"  Batch Size: Full Batch")
print(f"\nFinal Results:")
print(f"  Train Accuracy: {history['train_acc'][-1]:.2f}%")
print(f"  Test Accuracy: {history['test_acc'][-1]:.2f}%")
print(f"  Train Loss: {history['train_loss'][-1]:.4f}")
print(f"  Test Loss: {history['test_loss'][-1]:.4f}")
print(f"{'='*60}")
```
