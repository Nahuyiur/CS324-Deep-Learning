# Part III: Stochastic Gradient Descent

## Task 1: SGD with Default Parameters

Train the MLP using Stochastic Gradient Descent (batch_size=1) with default parameters.

```python
import matplotlib.pyplot as plt
from train_mlp_numpy import train, DNN_HIDDEN_UNITS_DEFAULT, LEARNING_RATE_DEFAULT, MAX_EPOCHS_DEFAULT, EVAL_FREQ_DEFAULT

# Train with SGD (batch_size=1) using default parameters
history_sgd = train(
    dnn_hidden_units=DNN_HIDDEN_UNITS_DEFAULT,
    learning_rate=LEARNING_RATE_DEFAULT,
    max_steps=MAX_EPOCHS_DEFAULT,
    eval_freq=EVAL_FREQ_DEFAULT,
    batch_size=1,  # SGD
    return_history=True
)

# Plot accuracy and loss curves for SGD
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(history_sgd['steps'], history_sgd['train_acc'], label='Train Accuracy', marker='o', markersize=3)
ax1.plot(history_sgd['steps'], history_sgd['test_acc'], label='Test Accuracy', marker='s', markersize=3)
ax1.set_xlabel('Training Steps')
ax1.set_ylabel('Accuracy (%)')
ax1.set_title('Training and Test Accuracy (SGD, batch_size=1)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(history_sgd['steps'], history_sgd['train_loss'], label='Train Loss', marker='o', markersize=3)
ax2.plot(history_sgd['steps'], history_sgd['test_loss'], label='Test Loss', marker='s', markersize=3)
ax2.set_xlabel('Training Steps')
ax2.set_ylabel('Loss')
ax2.set_title('Training and Test Loss (SGD, batch_size=1)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n{'='*60}")
print(f"Experiment Parameters (Default - SGD):")
print(f"  Hidden Units: {DNN_HIDDEN_UNITS_DEFAULT}")
print(f"  Learning Rate: {LEARNING_RATE_DEFAULT}")
print(f"  Max Steps: {MAX_EPOCHS_DEFAULT}")
print(f"  Eval Frequency: {EVAL_FREQ_DEFAULT}")
print(f"  Batch Size: 1 (SGD)")
print(f"\nFinal Results:")
print(f"  Train Accuracy: {history_sgd['train_acc'][-1]:.2f}%")
print(f"  Test Accuracy: {history_sgd['test_acc'][-1]:.2f}%")
print(f"  Train Loss: {history_sgd['train_loss'][-1]:.4f}")
print(f"  Test Loss: {history_sgd['test_loss'][-1]:.4f}")
print(f"{'='*60}")
```

## Task 2: Analysis of Batch Size Influence

Compare the effect of different batch sizes on training performance, from 1 to relatively large values.

```python
import matplotlib.pyplot as plt
from train_mlp_numpy import train, DNN_HIDDEN_UNITS_DEFAULT, LEARNING_RATE_DEFAULT, MAX_EPOCHS_DEFAULT, EVAL_FREQ_DEFAULT

# Test different batch sizes
batch_sizes = [1, 8, 32, 64, 128, None]  # None = full batch
histories = {}

print("Training with different batch sizes...\n")
for bs in batch_sizes:
    bs_name = 'Full Batch' if bs is None else f'Batch {bs}'
    print(f"Training with {bs_name}...")
    
    histories[bs_name] = train(
        dnn_hidden_units=DNN_HIDDEN_UNITS_DEFAULT,
        learning_rate=LEARNING_RATE_DEFAULT,
        max_steps=MAX_EPOCHS_DEFAULT,
        eval_freq=EVAL_FREQ_DEFAULT,
        batch_size=bs,
        return_history=True
    )
    print()

# Plot comparison of different batch sizes
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Training accuracy
for name, hist in histories.items():
    ax1.plot(hist['steps'], hist['train_acc'], label=name, marker='o', markersize=2)
ax1.set_xlabel('Training Steps')
ax1.set_ylabel('Accuracy (%)')
ax1.set_title('Training Accuracy - Different Batch Sizes')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Test accuracy
for name, hist in histories.items():
    ax2.plot(hist['steps'], hist['test_acc'], label=name, marker='s', markersize=2)
ax2.set_xlabel('Training Steps')
ax2.set_ylabel('Accuracy (%)')
ax2.set_title('Test Accuracy - Different Batch Sizes')
ax2.legend()
ax2.grid(True, alpha=0.3)

# Training loss
for name, hist in histories.items():
    ax3.plot(hist['steps'], hist['train_loss'], label=name, marker='o', markersize=2)
ax3.set_xlabel('Training Steps')
ax3.set_ylabel('Loss')
ax3.set_title('Training Loss - Different Batch Sizes')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Test loss
for name, hist in histories.items():
    ax4.plot(hist['steps'], hist['test_loss'], label=name, marker='s', markersize=2)
ax4.set_xlabel('Training Steps')
ax4.set_ylabel('Loss')
ax4.set_title('Test Loss - Different Batch Sizes')
ax4.legend()
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Summary table
print(f"\n{'='*80}")
print(f"{'Batch Size':<15} {'Train Acc':<15} {'Test Acc':<15} {'Train Loss':<15} {'Test Loss':<15}")
print(f"{'='*80}")
for name, hist in histories.items():
    print(f"{name:<15} {hist['train_acc'][-1]:>13.2f}% {hist['test_acc'][-1]:>13.2f}% "
          f"{hist['train_loss'][-1]:>13.4f}  {hist['test_loss'][-1]:>13.4f}")
print(f"{'='*80}")
```

## Analysis and Observations

### Key Findings:

**1. Convergence Speed:**
- **Small batch sizes (1-8)**: Converge faster in early epochs due to more frequent weight updates
- **Large batch sizes (64-128, Full Batch)**: Slower convergence than small batch sizes

**2. Training Stability:**
- **SGD (batch_size=1)**: More noisy updates, but often escapes local minima better
- **Mini-batch (8-64)**: Good balance between speed and stability
- **Full batch**: Smooth but very slow with default learning rate

**3. Final Performance:**
- SGD and small batch sizes typically achieve better test accuracy
- Larger batch sizes may require more epochs or higher learning rates to achieve similar performance

**4. Computational Efficiency:**
- SGD: More weight updates per epoch
- Full batch: Only 1 update per epoch
- Mini-batch: Good compromise between update frequency and computational efficiency（trade-off）
